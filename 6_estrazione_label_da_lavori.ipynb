```python
import pandas as pd
import spacy
nlp = spacy.load("en_core_web_lg")
```

```python
df = pd.read_csv("data/data_job_posts2.csv")
df.head()
```

```python
def get_labels(parola_iniziale, text):
    text = list(dict.fromkeys(text.split()))
    testo = ' '.join(text)
    doc = nlp(testo)

    # Trova le parole più simili alla parola iniziale nel testo
    parole_simili = []
    for token in doc:
        # Calcola la similarità tra la parola iniziale e ogni altra parola nel testo
        similarita = nlp(parola_iniziale).similarity(token)
        parole_simili.append((token.text, similarita))

    # Ordina le parole in base alla similarità e ottieni le tre parole più simili
    parole_simili = sorted(parole_simili, key=lambda x: x[1], reverse=True)[1:4]
    
    vet_parole = []
    vet_similarita = []

    # Stampa le tre parole più simili
    print(f"Le tre parole più simili a '{parola_iniziale}' nel testo sono:")
    for parola, similarita in parole_simili:
        vet_parole.append(parola)
        vet_similarita.append(similarita)
        print(f"{parola}: {similarita}")
    print("------------------------")
    
    return vet_parole, vet_similarita
```

```python
parole = []
similarita = []
for index, row in df.iterrows(): 
    vet_parole, vet_similarita = get_labels(row["Title"], row["JobDescription"])
    parole.append(vet_parole)
    similarita.append(vet_similarita)
```

```python
df_parole = pd.DataFrame(parole, columns=['prima_label', 'seconda_label', 'terza_label'])
df_similarita = pd.DataFrame(similarita, columns=['prima_similarita', 'seconda_similarita', 'terza_similarita'])

df_completo = pd.concat([df, df_parole, df_similarita], axis=1)
```

```python
df_completo.to_csv("data/df_lavori_completo.csv", index = False)
```

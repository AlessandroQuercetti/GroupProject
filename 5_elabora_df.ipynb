```python
import pandas as pd
import spacy
nlp = spacy.load("en_core_web_lg")
```

```python
df = pd.read_csv("data/df_persone.csv")
df.head()
```

```python
df = df.fillna("")
```

estraiamo le parole chiavi dal testo, ma non va bene perchè restituisce tutte le parole contenute nel testo

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# Utilizza il TF-IDF per estrarre le etichette dalla descrizione del lavoro
tfidf_vectorizer = TfidfVectorizer(stop_words='english')  # Rimuovi parole comuni in inglese
tfidf_matrix = tfidf_vectorizer.fit_transform(df['prep_summary'])

# Trova le parole chiave per ogni descrizione del lavoro
parole_chiave = []
for i, row in enumerate(df.itertuples()):
    feature_names = tfidf_vectorizer.get_feature_names_out()
    parole_chiave_descrizione = [feature_names[idx] for idx in tfidf_matrix[i].indices]
    parole_chiave.append(parole_chiave_descrizione)

# Aggiungi le parole chiave al DataFrame originale
df['parole_chiave'] = parole_chiave

# Stampa i risultati
df
```

facciamo il confronto tra il ruolo e tra le parole contenute nella descrizione, troviamo le 3 più simili al ruolo. 
Lo facciamo sia per prep_skills che per prep_summary.

```python
def get_labels(parola_iniziale, text):
    text = list(dict.fromkeys(text.split()))
    testo = ' '.join(text)
    doc = nlp(testo)

    # Trova le parole più simili alla parola iniziale nel testo
    parole_simili = []
    for token in doc:
        # Calcola la similarità tra la parola iniziale e ogni altra parola nel testo
        similarita = nlp(parola_iniziale).similarity(token)
        parole_simili.append((token.text, similarita))

    # Ordina le parole in base alla similarità e ottieni le tre parole più simili
    parole_simili = sorted(parole_simili, key=lambda x: x[1], reverse=True)[1:4]
    
    vet_parole = []
    vet_similarita = []

    # Stampa le tre parole più simili
    print(f"Le tre parole più simili a '{parola_iniziale}' nel testo sono:")
    for parola, similarita in parole_simili:
        vet_parole.append(parola)
        vet_similarita.append(similarita)
        print(f"{parola}: {similarita}")
    print("------------------------")
    
    return vet_parole, vet_similarita
```

lo faccio per summary

```python
parole_summary = []
similarita_summary = []
for index, row in df.iterrows(): 
    vet_parole, vet_similarita = get_labels(row["ruolo"], row["prep_summary"])
    parole_summary.append(vet_parole)
    similarita_summary.append(vet_similarita)
```

lo faccio per skills

```python
parole_skills = []
similarita_skills = []
for index, row in df.iterrows(): 
    vet_parole, vet_similarita = get_labels(row["ruolo"], row["prep_skills"])
    parole_skills.append(vet_parole)
    similarita_skills.append(vet_similarita)
```

creo colonne nel df e lo salvo

```python
df_summary_parole = pd.DataFrame(parole_summary, columns=['prima_label_summary', 'seconda_label_summary', 'terza_label_summary'])
df_summary_similarita = pd.DataFrame(similarita_summary, columns=['prima_similarita_summary', 'seconda_similarita_summary', 'terza_similarita_summary'])
df_skills_parole = pd.DataFrame(parole_skills, columns=['prima_label_skills', 'seconda_label_skills', 'terza_label_skills'])
df_skills_similarita = pd.DataFrame(similarita_skills, columns=['prima_similarita_skills', 'seconda_similarita_skills', 'terza_similarita_skills'])


# Concatena il DataFrame temporaneo con il DataFrame esistente lungo l'asse delle colonne (axis=1)
df_completo = pd.concat([df, df_summary_parole, df_summary_similarita, df_skills_parole, df_skills_similarita], axis=1)
```

```python
import pandas as pd
import spacy
nlp = spacy.load("en_core_web_lg")
```

```python
df = pd.read_csv("data/df_persone.csv")
df.head()
```

```python
def filtra(text):
    try:
        text = text.replace('(', ' ').replace(')', ' ').replace('[', ' ').replace(']', ' ').replace('{', ' ').replace('}', ' ').strip()
        doc = nlp(text)
        filtered_tokens = []
        for token in doc:
            if token.pos_ not in ["SPACE", "PUNCT", "X", "SYM"]:
                if not token.is_stop: 
                    filtered_tokens.append(token.lemma_)
        return ' '.join(e for e in filtered_tokens)
    except:
        print("errore nella funzione: filtra")
        return " "
    
def deformatta(text):
    try:
        text = text.replace('\n', ' ').replace('\t', ' ').replace('\r', ' ').casefold()
        return text
    except:
        print("errore nella funzione: filtra1")
        return " "
```

```python
def get_frasi(text): 
    doc = nlp(deformatta(text))
    summary_frasi = [sent.text.replace('summary', '') for sent in doc.sents if "summary" in sent.text]
    skills_frasi = [sent.text.replace('skills', '') for sent in doc.sents if "skills" in sent.text]
    #vedi anche se c'è da togliere la parola collegata al lavoro, es. accountant ecc...
    
    print("Summary:")
    for frase in summary_frasi:
        print(frase)
        
    print("Skills:")
    for frase in skills_frasi:
        print(frase)
        
    print("----------")
    return summary_frasi, skills_frasi
```

```python
#vec_summary = df["testo"].map(lambda x: get_summary(x))
#vec_skills = df["testo"].map(lambda x: get_skills(x))
summary_skills = df["testo"].map(lambda x: get_frasi(x))
```

```python
summary = []
skills = []
for coppia in summary_skills: 
    
    appoggio= ""
    for el in coppia[0]: appoggio = appoggio + el
    summary.append(appoggio)
    
    appoggio= ""
    for el in coppia[1]: appoggio = appoggio + el
    skills.append(appoggio)

```

```python
df["summary"] = summary
df["skills"] = skills
```

```python
df["prep_summary"] = df["summary"].map(lambda x: filtra(x))
df["prep_skills"] = df["skills"].map(lambda x: filtra(x))
```

```python
df.to_csv("data/df_persone.csv", index=False)
```

```python
import pandas as pd
import spacy
nlp = spacy.load("en_core_web_lg")
```

```python
df_persone = pd.read_csv("df_persone_completo.csv")
df_persone.head()
```

```python
df_lavori = pd.read_csv("df_lavori_completo.csv")
df_lavori.head()
```

```python
def calcola_sim(testo1, testo2): 
    doc1 = nlp(testo1)
    doc2 = nlp(testo2)
    return doc1.similarity(doc2)
```

```python
i = 0
label_summary = df_persone["prima_label_summary"][i] + " " + df_persone["seconda_label_summary"][i] + " " + df_persone["terza_label_summary"][i]
label_skills = df_persone["prima_label_skills"][i] + " " + df_persone["seconda_label_skills"][i] + " " + df_persone["terza_label_skills"][i]
label_persona = label_summary + " " + label_skills
label_lavoro = df_lavori["prima_label"][i] + " " + df_lavori["seconda_label"][i] + " " + df_lavori["terza_label"][i]
```

```python
print(calcola_sim(label_persona, label_lavoro))
```

```python
import random 
random_numbers = random.sample(range(0, len(df_persone["ruolo"])), 50)
```

```python
def trova_migliore(label_persona):
    max_sim=-1
    max_index=-1
    for index, row in df_lavori.iterrows(): 
        label_lavoro = row["prima_label"] + " " + row["seconda_label"] + " " + row["terza_label"]
        sim=calcola_sim(label_lavoro,label_persona)
        if sim>max_sim:
            max_sim=sim
            max_index=index
   
    return max_sim,max_index     
    
```

```python
print(trova_migliore('department financial reporting'))
```

```python
 for index, row in df_lavori.iterrows(): 
        print(index)
        print( row["prima_label"] + " " + row["seconda_label"] + " " + row["terza_label"])
        
```

```python
df_lavori['prima_label']= df_lavori['prima_label'].fillna("a")
df_lavori['terza_label']= df_lavori['terza_label'].fillna("a")
df_lavori['seconda_label']= df_lavori['seconda_label'].fillna("a")
```

```python
vet_index_persone=[]
vet_index_lavori=[]
vet_similarita=[]
for i in random_numbers:
    label_summary = df_persone["prima_label_summary"][i] + " " + df_persone["seconda_label_summary"][i] + " " + df_persone["terza_label_summary"][i]
    label_skills = df_persone["prima_label_skills"][i] + " " + df_persone["seconda_label_skills"][i] + " " + df_persone["terza_label_skills"][i]
    label_persona = label_summary + " " + label_skills
    similarita,indice=trova_migliore(label_persona)
    vet_index_persone.append(i)
    vet_similarita.append(similarita)
    vet_index_lavori.append(indice)
```

```python
df_persone['prima_label_summary']= df_persone['prima_label_summary'].fillna("a")
df_persone['terza_label_summary']= df_persone['terza_label_summary'].fillna("a")
df_persone['seconda_label_summary']= df_persone['seconda_label_summary'].fillna("a")
df_persone['prima_label_skills']= df_persone['prima_label_skills'].fillna("a")
df_persone['terza_label_skills']= df_persone['terza_label_skills'].fillna("a")
df_persone['seconda_label_skills']= df_persone['seconda_label_skills'].fillna("a")
```

```python
data = {'Indice Persone': vet_index_persone, 'Indice Lavori': vet_index_lavori, 'Similarita': vet_similarita}
```

```python
df=pd.DataFrame(data)
```

```python
df.to_csv('similarita_prova.csv', index=False)
```
